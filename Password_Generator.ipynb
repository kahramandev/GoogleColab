```python
import random
import string
def password_generator():
    print("Hi, welcome to Password Generator!")
    length = int(input("Please, enter length of password: "))
    lowers = string.ascii_lowercase
    uppers = string.ascii_uppercase
    digits = string.digits
    symbols = string.punctuation
    all = lowers + uppers + digits + symbols
    password = ''.join(random.sample(all, length))
    print(password)
    return f"Length of your password is: {len(password)}"
password_generator()
```

Bu Python kodu, kullanıcının belirlediği uzunlukta, tamamen rastgele ve güvenli bir şifre oluşturan bir programdır. Kod; küçük harf, büyük harf, rakam ve özel karakterleri karıştırarak tahmin edilmesi zor şifreler üretir.

**Kodun Çalışma Mantığı**

**1. Modüllerin İçe Aktarılması (import)**
*   `import random`: Rastgele seçimler yapabilmek için Python'ın rastgelelik modülünü yükler.
*   `import string`: Harf, rakam ve sembol gibi hazır karakter kümelerini hazır olarak almak için kullanılır.

**2. Kullanıcıdan Uzunluk Alınması**
*   `length = int(input(...))`: Kullanıcıya bir mesaj gösterir ve şifrenin kaç karakterden oluşmasını istediğini sorar. Alınan metin türündeki girdi, `int()` fonksiyonu ile tam sayıya (integer) dönüştürülür.

**3. Karakter Havuzunun Oluşturulması (all)**

Kod, şifrede kullanılabilecek tüm karakterleri `string` modülünü kullanarak tek bir havuzda toplar:
*   **lowers**: Küçük harfler (a-z)
*   **uppers**: Büyük harfler (A-Z)

**4. Şifrenin Üretilmesi (random.sample)**
`random.sample(all, length)`: Oluşturulan all havuzundan, kullanıcının girdiği

*   length (uzunluk) adedi kadar birbirinden farklı ve benzersiz karakteri rastgele seçer. Bu fonksiyon geriye bir liste (örn: ['a', '5', '!']) döndürür.
*   `''.join(...)`: Listenin içindeki bu rastgele karakterleri aralarında boşluk bırakmadan birleştirerek tek bir metin (string) haline getirir.

**Önemli Bir Kısıtlama: random.sample Hatası**

Bu kodda teknik olarak küçük bir risk vardır: `random.sample` fonksiyonu benzersiz (tekrarlanmayan) karakterler seçer.
string modülündeki toplam karakter sayısı **94 adettir.** Eğer kullanıcı şifre uzunluğunu 94'ten büyük bir sayı girerse (örneğin 100), kod karakter havuzunda yeterli benzersiz karakter bulamayacağı için **ValueError hatası vererek çökecektir.**

Kodun çökmesini önlemek ve 94 karakterden daha uzun şifreler de üretebilmesini sağlamak için `random.sample` yerine `random.choices` fonksiyonunu kullanmalıyız.
`random.choices`, havuzdan seçtiği bir karakteri tekrar seçebilir (yerine koyarak seçme). Bu sayede hem şifredeki karakter çeşitliliği artar hem de kullanıcı ne kadar büyük bir sayı girerse girsin program hatasız çalışır.

```python
import random
import string

def password_generator():
    print("Hi, welcome to Password Generator!")

    # Kullanıcının hatalı veri (harf vb.) girmesine karşı basit bir kontrol
    try:
        length = int(input("Please, enter length of password: "))
        if length <= 0:
            print("Password length must be greater than 0!")
            return
    except ValueError:
        print("Please enter a valid number!")
        return

    lowers = string.ascii_lowercase
    uppers = string.ascii_uppercase
    digits = string.digits
    symbols = string.punctuation

    all_characters = lowers + uppers + digits + symbols

    # random.sample yerine random.choices kullanarak karakter tekrarına izin verdik:
    password = ''.join(random.choices(all_characters, k=length))

    print(f"Generated Password: {password}")
    return f"Length of your password is: {len(password)}"

password_generator()
```

# Neler Değişti?



*   `random.choices(all_characters, k=length)`: Bu fonksiyon sayesinde aynı harf veya sembol şifre içinde birden fazla kez geçebilir. Kullanıcı uzunluğu 150 bile girse program sorunsuz çalışır.
*   `try-except` Bloğu: Kullanıcı sayı yerine yanlışlıkla harf girerse (örneğin "on" yazarsa) programın çökmesi engellendi ve kullanıcıya uyarı mesajı gösterildi.

Şimdi şifrenin daha güvenli olması için **en az 1 büyük harf, 1 küçük harf ve 1 rakam içerme zorunluluğu** gibi kurallar da ekleyelim.

Şifremize belirli kurallar (en az 1 büyük harf, 1 küçük harf, 1 rakam ve 1 sembol) eklemek için **iki farklı yaklaşım** kullanabiliriz.


# 1. Yöntem: Kesin Garantili Yöntem (Önerilen)

Bu yöntem, kullanıcının istediği uzunluk ne olursa olsun (örneğin sadece 4 karakter bile girse) şifrenin içinde **kesinlikle en az birer adet** küçük harf, büyük harf, rakam ve sembol bulunmasını garanti eder. Geri kalan karakterleri rastgele tamamlar ve sırayı karıştırır.

```python
import random
import string

def secure_password_generator():
    print("Hi, welcome to Secure Password Generator!")

    try:
        length = int(input("Please, enter length of password (min 4): "))
        if length < 4:
            print("For a secure password, length must be at least 4!")
            return
    except ValueError:
        print("Please enter a valid number!")
        return

    lowers = string.ascii_lowercase
    uppers = string.ascii_uppercase
    digits = string.digits
    symbols = string.punctuation
    all_characters = lowers + uppers + digits + symbols

    # 1. Aşama: Her gruptan kesinlikle 1 tane karakter seçiyoruz (Zorunluluk)
    password_parts = [
        random.choice(lowers),
        random.choice(uppers),
        random.choice(digits),
        random.choice(symbols)
    ]

    # 2. Aşama: Geri kalan boşlukları tüm havuzdan rastgele dolduruyoruz
    remaining_length = length - 4
    if remaining_length > 0:
        password_parts += random.choices(all_characters, k=remaining_length)

    # 3. Aşama: İlk karakterlerin sırası belli olmasın diye tüm listeyi karıştırıyoruz
    random.shuffle(password_parts)

    # Listeyi metne dönüştürüyoruz
    password = ''.join(password_parts)

    print(f"Generated Password: {password}")
    return f"Length of your password is: {len(password)}"

secure_password_generator()
```

# 2. Yöntem: Kontrol Döngüsü Yöntemi (Doğal Dağılım)

Bu yöntem ise tamamen rastgele bir şifre üretir ve ardından şifreyi kontrol eder. Eğer şifre kurallara uymuyorsa (örneğin içinde hiç rakam yoksa), kurallara uyan bir şifre denk gelene kadar **yeniden şifre üretmeye devam eder.** Uzun şifrelerde çok hızlı çalışır.

```python
import random
import string

def check_password_generator():
    print("Hi, welcome to Conditional Password Generator!")

    try:
        length = int(input("Please, enter length of password (min 4): "))
        if length < 4:
            print("Length must be at least 4 to satisfy all conditions!")
            return
    except ValueError:
        print("Please enter a valid number!")
        return

    lowers = string.ascii_lowercase
    uppers = string.ascii_uppercase
    digits = string.digits
    symbols = string.punctuation
    all_characters = lowers + uppers + digits + symbols

    while True:
        # Rastgele şifre üret
        password = ''.join(random.choices(all_characters, k=length))

        # Koşulları kontrol et (any() fonksiyonu true/false döner)
        has_lower = any(char in lowers for char in password)
        has_upper = any(char in uppers for char in password)
        has_digit = any(char in digits for char in password)
        has_symbol = any(char in symbols for char in password)

        # Eğer tüm koşullar sağlanıyorsa döngüden çık
        if has_lower and has_upper and has_digit and has_symbol:
            break

    print(f"Generated Password: {password}")
    return f"Length of your password is: {len(password)}"

check_password_generator()
```

**Decorator (Dekoratör) Yaklaşımı**

Bu yaklaşım, şifre üretme mantığını bir "dekoratör" içine saklayarak, asıl fonksiyonu `(generate_password)` sadece şifreyi ekrana basmak veya işlemekle görevlendirir.

```python
import random
import string

def password_generator(length):
    def decorator(func):
        def wrapper(*args, **kwargs):
            password_characters = string.ascii_letters + string.digits + string.punctuation
            password = ''.join(random.choice(password_characters) for _ in range(length))
            return func(password, *args, **kwargs)
        return wrapper
    return decorator

@password_generator(length=int(input()))
def generate_password(password):
    print("Generated password:", password)

generate_password()
```

**Nasıl Çalışır ve Avantajı Nedir?**

Ayrılmış Sorumluluk: Şifre üretme mantığı ile şifrenin kullanıcıya gösterilmesi mantığı birbirinden tamamen ayrılmıştır. `@password_generator` dekoratörünü yarın bir gün bir "şifreyi veritabanına kaydetme" fonksiyonunun üzerine de koysanız yine çalışır.

**Kritik Hata (Tasarım Sorunu):**

Bu kodda çok ciddi bir mimari problem var: `@password_generator(length=int(input()))` satırı, kod satır satır okunup **belleğe yüklenirken (import/compile anında) çalışır**, fonksiyon çağrıldığında değil.


*   Programı çalıştırdığınız an, daha `generate_password()` satırına gelmeden sizden terminalde girdi bekler.
*   Eğer bu fonksiyonu bir döngü içinde 5 kez çağırmak isterseniz, her seferinde yeni girdi isteyemezsiniz; çünkü uzunluk kodu ilk yüklendiğinde bir kez belirlenmiştir.

**Generator (yield) Yaklaşımı**

Bu yaklaşım, Python'ın bellek dostu `yield` (Generator) yapısını kullanır. Şifreyi hemen üretip belleğe atmak yerine, ihtiyaç duyulduğu an üretilmesini sağlar.

```python
import random
import string

def password_generator(length, chars=string.ascii_letters + string.digits + string.punctuation):
  """
  Yields a random password of specified length.
  """
  for _ in range(1): # This loop ensures we only generate one password per call
    yield ''.join(random.choice(chars) for _ in range(length))

# Usage:
password_gen = password_generator(12)  # Create a generator object for a 12-char password
password = next(password_gen)       # Get the next (and only) password from the generator
print(password)
```

**Nasıl Çalışır ve Avantajı Nedir?**

*   **Tembel Değerlendirme (Lazy Evaluation)**: `password_generator(12)` dediğinizde şifre henüz üretilmez. Sadece üretmeye hazır bir nesne yaratılır. Ne zaman ki `next()` fonksiyonunu çağırırsınız, şifre tam o salisede üretilir.
*   **Geliştirilebilir Yapı**: Kodun içindeki `for _ in range(1)`: kısmı şu an tek şifre üretiyor ancak orayı `while True`: yaparsanız, her `next()` dediğinizde sonsuza kadar **yeni ve farklı şifreler üreten** bir "şifre fabrikası" elde edersiniz.
